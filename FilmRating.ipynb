```python
rating = pd.read_csv('ratings.csv')
rating.head()
```

```python
def movie_rating(rating):
    if rating <= 2:
          return 'Низкий'
    if 2 < rating <= 4:
          return 'Средний'
    else:
        return 'Высокий'

rating['class'] = rating['rating'].apply(movie_rating)

rating.head(10)
```
